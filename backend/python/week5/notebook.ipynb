### Fetch Products
Fetch product data from MongoDB.

### Create DataFrame
Convert queried data into a Pandas DataFrame for analysis.

````python
import pandas as pd

products = Product.objects()

data = []
for p in products:
    data.append({
        "name": p.name,
        "brand":p.brand,
        "price":p.price,
        "quantity": p.quantity
    })

df = pd.DataFrame(data)
df
````

### Stock Visualization
Plot product quantities to understand stock distribution.

````python
import matplotlib.pyplot as plt
import pandas as pd

plt.figure()
plt.bar(df["name"], df["quantity"])
plt.xlabel("Product")
plt.ylabel("Quantity")
plt.title("Stock levels")
plt.xticks(rotation=90)
plt.show()
````
