```python
import pandas as pd

products = Product.objects()

data = []
for p in products:
    data.append({
        "name": p.name,
        "brand":p.brand,
        "price":p.price,
        "quantity": p.quantity
    })

df = pd.DataFrame(data)
df
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame(data)
plt.figure()
plt.bar(df["name"], df["quantity"])
plt.xlabel("Product")
plt.ylabel("Quantity")
plt.title("Stock levels")
plt.xticks(rotation=90)
plt.show()
```

```python
import seaborn as sns
import pandas as pd

df = pd.DataFrame(data)
plt.figure()
sns.barplot(x="name", y="quantity", data=df)
plt.xticks(rotation=90)
plt.title("Stock levels")

plt.show()
```
